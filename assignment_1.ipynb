```python
import pandas as pd

df = pd.read_csv('/Combined_dataset.csv')

print("--- FIRST 5 ROWS ---")
print(df.head())

print("\n--- DATA SHAPE ---")
print(df.shape)

print("\n--- COLUMNS AND DATA TYPES ---")
print(df.info())
```

```python
df_cleaned = df.copy()
df_cleaned['discount'] = df_cleaned['discount'].fillna(0)
df_cleaned = df_cleaned.fillna('Unknown')

print("Missing values handled completely!")
```

```python
filtered_df = df_cleaned[df_cleaned['rating'] > 4.0]

selected_df = filtered_df[['product_id', 'title', 'rating', 'initial_price', 'discount', 'category']]

final_df = selected_df.drop_duplicates()

print(f"Original row count: {df.shape[0]}")
print(f"Cleaned row count: {final_df.shape[0]}")
```

```python
final_df = final_df.copy()
final_df['total_amount'] = final_df['initial_price']

final_df.to_csv('cleaned_shopping_dataset.csv', index=False)
print("Cleaned dataset file 'cleaned_shopping_dataset.csv' has been generated successfully!")
```
